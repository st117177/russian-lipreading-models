```python
def load_video(path, num_frames=24, size=96):
    cap = cv2.VideoCapture(str(path))

    frames = []
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames <= 0:
        cap.release()
        raise ValueError(f"Cannot read video: {path}")

    frame_indices = np.linspace(0, total_frames - 1, num_frames).astype(int)

    for frame_idx in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
        ok, frame = cap.read()

        if not ok:
            continue

        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        frame = cv2.resize(frame, (size, size))
        frame = frame / 255.0

        frames.append(frame)

    cap.release()

    if len(frames) == 0:
        raise ValueError(f"No frames loaded: {path}")

    while len(frames) < num_frames:
        frames.append(frames[-1])

    video = np.stack(frames)

    video = torch.tensor(video, dtype=torch.float32)
    video = video.permute(3, 0, 1, 2)

    return video
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for batch_x, batch_y in loader:
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()

        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * batch_x.size(0)

        preds = outputs.argmax(dim=1)
        correct += (preds == batch_y).sum().item()
        total += batch_y.size(0)

    avg_loss = total_loss / total
    accuracy = correct / total

    return avg_loss, accuracy
```

```python
def evaluate(model, loader, criterion, device):
    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        for batch_x, batch_y in loader:
            batch_x = batch_x.to(device)
            batch_y = batch_y.to(device)

            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)

            total_loss += loss.item() * batch_x.size(0)

            preds = outputs.argmax(dim=1)
            correct += (preds == batch_y).sum().item()
            total += batch_y.size(0)

    avg_loss = total_loss / total
    accuracy = correct / total

    return avg_loss, accuracy
```

```python
model = Simple3DCNN(num_classes=len(vocab)).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
print(train_loss, train_acc)

val_loss, val_acc = evaluate(model, val_loader, criterion, device)
print(val_loss, val_acc)
```

```python
# Сначала просто считаем loss и accuracy на validation.
# Это та же функция evaluate, которую мы уже написали.
val_loss, val_acc = evaluate(model, val_loader, criterion, device)

# Сюда будем складывать предсказания модели.
all_preds = []

# Сюда будем складывать правильные ответы.
all_labels = []

# Переводим модель в режим проверки.
# В этом режиме модель НЕ обучается.
model.eval()

# Говорим PyTorch: градиенты считать не надо.
# Это быстрее и экономит память, потому что мы сейчас только проверяем модель.
with torch.no_grad():

    # Идем по validation batch за batch.
    # batch_x = пачка видео
    # batch_y = правильные классы для этих видео
    for batch_x, batch_y in val_loader:

        # Переносим видео на GPU или CPU, смотря что лежит в device.
        batch_x = batch_x.to(device)

        # Переносим правильные ответы туда же.
        batch_y = batch_y.to(device)

        # Модель делает предсказание.
        # outputs имеет форму [batch_size, num_classes].
        # Например [8, 10]: 8 видео, 10 чисел для каждого видео.
        outputs = model(batch_x)

        # Берем индекс самого большого числа.
        # Это и есть предсказанный класс.
        # Например если самое большое число у класса 0, модель сказала "есть".
        preds = outputs.argmax(dim=1)

        # Переносим preds с GPU на CPU,
        # превращаем в numpy,
        # потом в обычный Python list,
        # и добавляем в общий список all_preds.
        all_preds.extend(preds.cpu().numpy().tolist())

        # То же самое делаем с правильными ответами.
        all_labels.extend(batch_y.cpu().numpy().tolist())

# Counter считает, сколько раз встретилось каждое число.
# Например Counter({0: 105}) значит: модель 105 раз предсказала класс 0.
from collections import Counter

print("preds:", Counter(all_preds))
print("labels:", Counter(all_labels))

# Теперь печатаем предсказания не номерами, а словами.
print("pred words:")
for idx, count in Counter(all_preds).most_common():
    print(idx_to_word[idx], count)

# И печатаем реальные правильные слова на validation.
print("true words:")
for idx, count in Counter(all_labels).most_common():
    print(idx_to_word[idx], count)
```

```python
class_weights = []

# Идем по vocab в правильном порядке классов:
# 0 класс, 1 класс, 2 класс...
for word in vocab:
    count = train_counts[word]

    # Чем меньше примеров слова, тем больше вес.
    weight = 1.0 / count

    class_weights.append(weight)

class_weights = torch.tensor(class_weights, dtype=torch.float32)

# Нормализуем веса, чтобы средний вес был примерно 1.
class_weights = class_weights / class_weights.mean()

class_weights = class_weights.to(device)

print(class_weights)
```

```python
criterion = nn.CrossEntropyLoss(weight=class_weights)
```
